```python
# 2. Automatically locate the dataset and copy it to working directory
import os

input_dir = "/kaggle/input"
dataset_name = None

# Find the dataset folder
if os.path.exists(input_dir):
    folders = os.listdir(input_dir)
    if folders:
        dataset_name = folders[0]
        print(f"Found dataset: {dataset_name}")
        
        src_path = os.path.join(input_dir, dataset_name)
        dst_path = "/kaggle/working/"
        
        # Copy everything
        os.system(f"cp -r {src_path}/. {dst_path}")
        print(f"Successfully copied {src_path} to {dst_path}")
    else:
        print("ERROR: No dataset found in /kaggle/input/. Did you attach the zip file?")
else:
    print("ERROR: /kaggle/input/ does not exist.")
```
